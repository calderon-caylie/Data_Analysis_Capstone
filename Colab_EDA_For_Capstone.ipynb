```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/Caylie-Mackay/Raw_CSV/refs/heads/main/bike_sales_100k.csv')
df.head()
```

```python
df.info()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.info()
```

This dataset does not have any duplicates or nulls.

```python
df.describe()
```

Some noteable descriptive statistics are the average price of a bike and age of a customer. The min and max of the price also helps us to know the price range of the products.

```python
Male_vs_female = df.groupby('Customer_Gender')['Customer_Gender'].count().reset_index(name='Count')
Male_vs_female
```

```python
df['Transaction Total'] = df['Quantity'] * df['Price']
df.head()
```

```python
df['Date'] = pd.to_datetime(df['Date'], format='%d-%m-%Y')
df.head()
```

```python
df['Year'] = df['Date'].dt.year
avg_total_by_year = df.groupby('Year')['Transaction Total'].mean().reset_index()
avg_total_by_year = avg_total_by_year.round(2)
avg_total_by_year
```

```python
Sales_Goal = avg_total_by_year.copy()
Sales_Goal['Transaction Total'] = Sales_Goal['Transaction Total'] + 100
Sales_Goal
```

```python
employee_sales_by_year = df.groupby(['Year', 'Salesperson_ID'])['Transaction Total'].mean().reset_index()
employee_sales_by_year
```

```python
employee_sales_by_year_2021 = employee_sales_by_year[employee_sales_by_year['Year'] == 2021].copy()

sales_goal_2021 = Sales_Goal[Sales_Goal['Year'] == 2020]['Transaction Total'].iloc[0]
employee_sales_by_year_2021['Met Sales Goal'] = np.where(
    employee_sales_by_year_2021['Transaction Total'] >= sales_goal_2021,
    'Goal Met',
    'Under Goal'
)
employee_sales_by_year_2021.head()
```

```python
employee_sales_by_year_2021.value_counts('Met Sales Goal')
```

```python
employee_sales_by_year_2022 = employee_sales_by_year[employee_sales_by_year['Year'] == 2022].copy()

sales_goal_2022 = Sales_Goal[Sales_Goal['Year'] == 2021]['Transaction Total'].iloc[0]
employee_sales_by_year_2022['Met Sales Goal'] = np.where(
    employee_sales_by_year_2022['Transaction Total'] >= sales_goal_2022,
    'Goal Met',
    'Under Goal'
)
employee_sales_by_year_2022.head()
```

```python
employee_sales_by_year_2022.value_counts('Met Sales Goal')
```

```python
employee_sales_by_year_2023 = employee_sales_by_year[employee_sales_by_year['Year'] == 2023].copy()

sales_goal_2023 = Sales_Goal[Sales_Goal['Year'] == 2022]['Transaction Total'].iloc[0]
employee_sales_by_year_2023['Met Sales Goal'] = np.where(
    employee_sales_by_year_2023['Transaction Total'] >= sales_goal_2023,
    'Goal Met',
    'Under Goal'
)
employee_sales_by_year_2023.head()
```

```python
employee_sales_by_year_2023.value_counts('Met Sales Goal')
```

```python
employee_sales_by_year_2024 = employee_sales_by_year[employee_sales_by_year['Year'] == 2024].copy()

sales_goal_2024 = Sales_Goal[Sales_Goal['Year'] == 2023]['Transaction Total'].iloc[0]
employee_sales_by_year_2024['Met Sales Goal'] = np.where(
    employee_sales_by_year_2024['Transaction Total'] >= sales_goal_2024,
    'Goal Met',
    'Under Goal'
)
employee_sales_by_year_2024.head()
```

```python
employee_sales_by_year_2024.value_counts('Met Sales Goal')
```

```python
all_goals = pd.concat([
    employee_sales_by_year_2021[['Year', 'Salesperson_ID', 'Met Sales Goal']],
    employee_sales_by_year_2022[['Year', 'Salesperson_ID', 'Met Sales Goal']],
    employee_sales_by_year_2023[['Year', 'Salesperson_ID', 'Met Sales Goal']],
    employee_sales_by_year_2024[['Year', 'Salesperson_ID', 'Met Sales Goal']]
])
```

```python
employee_sales_by_year = pd.merge(
    employee_sales_by_year,
    all_goals,
    on=['Year', 'Salesperson_ID'],
    how='left'
)

employee_sales_by_year
```

```python
employee_sales_by_year = employee_sales_by_year[
    employee_sales_by_year['Year'] != 2020
]
employee_sales_by_year.head()
```
